#### 拼接

```python
b = torch.ones([2,2])
c = torch.zeros([2,3])
torch.cat([b,c],dim=1)
# 要在哪个维度拼接，就要保证除了该维度以外的
```

```python
torch.chunk(c,chunks=2,dim=1)
```

```python
d =torch.arange(3, 12).view(3, 3)
d
```

```python
index = torch.tensor([[2, 1, 0]])
e = d.gather(0, index)
e
# 相当于，dim=0那维如index排列，其他维度还是01234的顺序
# 最后取出来的就是[2,0][1,1][0,2]位置的数
```

```python
a = torch.arange(4)
torch.reshape(a, [2,2])
```

```python
torch.reshape(a,[-1,2])
# -1表示未知数量
```

```python
src = torch.arange(1,11).reshape([2,5])
src
```

```python
index = torch.tensor([[0,1,2,0]])
torch.zeros(3,5,dtype=src.dtype).scatter_(0,index,src)
# 类似gather，就是把src的元素按照index和dim来写入目标矩阵
# 有下划线就是原位操作
```

```python
a = torch.arange(10).view([5,2])
a
```

```python
torch.split(a,[1,4])
# 切分成有多少个元素的数组
```

```python
a = torch.reshape(a, [5,1,2])
a = a.squeeze()
# a = torch.squeeze(a)
a.shape
# 去掉为1的维度
# 也可以指定dim去掉 特定为1的dim
```

```python
torch.unsqueeze(a,0).shape
```

```python
b = torch.arange(10,20).view([5,2])
b
```

```python
torch.stack([a,b]).shape
# 默认按照dim=0
```

```python
src =torch.arange(4,10).view([2,3])
torch.take(src, torch.tensor([0,2,5]))
# take相当于把src平铺成一维，然后按index取src的元素
```

```python
x = torch.tensor([[1,2],[3,4]])
torch.tile(x,(2,2))
# 复制
```

```python
x = torch.arange(4,10).view([2,3])
torch.transpose(x, 0 ,1).shape 
# 交换维度——转置
```

```python
torch.unbind(torch.arange(1,10).view([3,3]))
# 切片
```

```python
x= torch.randn(3,2)
y = torch.ones_like(x)
torch.where(x>0,x,y)
```
