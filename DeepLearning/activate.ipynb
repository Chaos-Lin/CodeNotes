### softmax
 
并非是一个线性的，如果数据方差越大会导致其最后softmax的结果相差值非常大

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
alpha1 = 0.1
alpha2 = 10
score = torch.randn(5)
prob1 = F.softmax(score * alpha1 , -1)
prob2 = F.softmax(score * alpha2 , -1)

# 什么叫雅可比矩阵啊


print(prob1)
print(prob2)
```
