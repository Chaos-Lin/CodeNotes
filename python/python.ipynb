# decorator
function
函数是一个object，只是一个callable
可以作为参数被传进，也可以被return

decorator
本质上也是一个callable
我们通常默认decorator是一个参数是函数，返回值也是函数的函数

```python
def dec(f):
    pass

@dec
def double(x):
    return x * 2

double = dec(double)

# 上述两个定义一致
```

```python
import time

def timeit(f):
    def wrapper(x):
        start = time.time()
        print('d')
        ret = f(x)
        # 被装饰的函数调用之地
        print('a')
        return ret*2
        # 最终的返回值
    return wrapper

@timeit
def my_func(x):
    print('b')
    time.sleep(x)
    print('c')
    return 2 * x
print(my_func(1))
```

```python
print('二者等价')

def my_func1(x):
    print('b')
    time.sleep(x)
    print('c')
    return 2 * x

my_func2 = timeit(my_func1)
# 此时便是一个封装好的wrapper函数了
print(my_func2(1))
```

```python
print('为曾被装饰的原函数')
print(my_func1(1))
```

*args就是就是传递一个可变参数列表给函数实参，这个参数列表的数目未知，甚至长度可以为0。

```python
def test_args(first, *args):
    print('Required argument: ', first)
    print(type(args))
    for v in args:
        print ('Optional argument: ', v)

test_args(1, 2, 3, 4)
```

# **kwargs
将一个可变的关键字参数的字典传给函数实参，同样参数列表长度可以为0或为其他值。

```python
def test_kwargs(first, *args, **kwargs):
   print('Required argument: ', first)
   print(type(kwargs))
   for v in args:
      print ('Optional argument (args): ', v)
   for k, v in kwargs.items():
      print ('Optional argument %s (kwargs): %s' % (k, v))

test_kwargs(1, 2, 3, 4, k1=5, k2=6)
```
