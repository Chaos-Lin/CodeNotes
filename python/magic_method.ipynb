# 魔术方法 Special Method

## new & init

```python
class A:
    def __new__(cls):
        print("__new__")
        return super().__new__(cls)
    # 从cls 建立 obj
    # 比如想要建立一个类的唯一对象，那么在new时需要判断是否有其他obj被建立了

    def __init__(self):
        print("__init__")
    # 初始化一个obj
o = A()
```

## del

```python
class A:
    def __del__(self):
        print("__del__")

o = A()
x = o
del x
# 这个关键字和__del__无关，只是单纯地使之少一个引用

print("finish")
```

## repr & str

```python
class A:
    def __repr__(self):
        return "<A>"
    # 更详细

    
    def __str__(self):
        return "<A> str"
    # 注重可读性
    # 未定义则直接调用repr

print(repr(A()))
print(A()) # 默认维str
print(str(A()))
```

## format

```python
class A:
    def __format__(self, format_spec: str) -> str:
        if format_spec == "x":
            return "0xA"
        return "<A>"
    
print("{}".format(15))
print("{:b}".format(15))
print("{:x}".format(15))

print(f"{15:b}")
print(f"{15:x}")

print(f"{A()}")
print(f"{A():x}")
```

## bytes

```python
class A:
    def __bytes__(self):
        print("__bytes__")
        return bytes([0,1])
print(bytes(A()))
```

## rich comparision
== != > < >= <=

当没有去写一个类的比较逻辑时，相等则调用is

eq ne gt lt gr le

小于等于 并不为 小于或等于


```python
class Date:
    def __init__(self, year, month, date):
        self.year = year 
        self.month = month
        self.date = date

    def __eq__(self, other):
        print("__eq__")
        return (self.year == other.year and
                self.month == other.month and
                self.date == other.date)
    def __ne__(self, other):
        print("__ne__")
        return (self.year != other.year or
                self.month != other.month or
                self.date != other.date)
    
    def __gt__(self, other):
        print("__gt__")
        if self.year > other.year:
            return True
        if self.year== other.year:
            if self.month > other.month:
                return True
            if self.month == self.month:
                return self.date > other.date

x = Date(2022, 2, 22)
y = Date(2022, 2, 22)

print(x == y)
print(x != y)
# x.__ne__(y)
# 没有定义!=时 会对==取反

print(x > y)
```

## hash
有默认的hash函数，但是如果定义了eq函数，默认的hash函数就会被删掉

相等对象必须返回相同的hash值，并且返回值必须是int

```python
class Date:
    def __init__(self, year, month, date):
        self.year = year 
        self.month = month
        self.date = date

    def __eq__(self, other):
        print("__eq__")
        return (self.year == other.year and
                self.month == other.month and
                self.date == other.date)
    
    def __hash__(self):
        return hash((self.year, self.month, self.date))
    
x = Date(2022, 2, 22)
y = Date(2022, 2, 22)
print(hash(x))
print(hash(y))
```

## bool

```python
class Date:
    def __init__(self, year, month, date):
        self.year = year 
        self.month = month
        self.date = date
    def __bool__(self):
        print("__bool__")
        return False
    # 如果不定义，则默认为真

x = Date(2022, 2, 22)
print(bool(x))
```

## Attribute
